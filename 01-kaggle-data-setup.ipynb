```python
## Step 2: Set Up Dataset Paths
import os

# List available input datasets
print("Available input datasets:")
for d in os.listdir('/kaggle/input'):
    print(f"  /kaggle/input/{d}")

# Set dataset path 
DATASET_PATH = '/kaggle/input/datasets/shubhamgoel27/dermnet' 
# Working directory for outputs
OUTPUT_PATH = '/kaggle/working'
os.makedirs(OUTPUT_PATH, exist_ok=True)

print(f"\nDataset Path: {DATASET_PATH}")
print(f"Output Path:  {OUTPUT_PATH}")
print("\n✓ Paths configured successfully!")
```

```python
## Step 3: Explore Dataset Structure

import os
from collections import defaultdict

def explore_dataset_structure(base_path):
    print(f"\n{'='*60}")
    print(f"DATASET STRUCTURE: {base_path}")
    print(f"{'='*60}\n")

    train_path = os.path.join(base_path, 'train')
    test_path  = os.path.join(base_path, 'test')

    for folder_type, folder_path in [('TRAIN', train_path), ('TEST', test_path)]:
        if os.path.exists(folder_path):
            print(f"\n📁 {folder_type} FOLDER:")
            print(f"   Path: {folder_path}")

            classes = [d for d in os.listdir(folder_path)
                      if os.path.isdir(os.path.join(folder_path, d))]
            print(f"   Number of classes: {len(classes)}")

            class_counts = {}
            for class_name in sorted(classes):
                class_path = os.path.join(folder_path, class_name)
                count = len([f for f in os.listdir(class_path)
                             if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
                class_counts[class_name] = count

            print(f"\n   Class Distribution:")
            for class_name, count in sorted(class_counts.items(), key=lambda x: x[1], reverse=True):
                print(f"      • {class_name}: {count} images")

            print(f"\n   Total images: {sum(class_counts.values())}")
        else:
            print(f"\n⚠️  {folder_type} folder not found at {folder_path}")

    print(f"\n{'='*60}\n")

explore_dataset_structure(DATASET_PATH)
```

```python
## Step 4: Save Configuration for Next Notebooks

import json

# Save config to working directory
paths_config = {
    'DATASET_PATH': DATASET_PATH,
    'OUTPUT_PATH': OUTPUT_PATH
}

config_path = os.path.join(OUTPUT_PATH, 'paths_config.json')
with open(config_path, 'w') as f:
    json.dump(paths_config, f, indent=4)

print(f"✓ Configuration saved to: {config_path}")
print("\nReady for Notebook 2!")
```
